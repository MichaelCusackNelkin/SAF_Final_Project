#### Hybrid filler

```python
SIDE = "BUY"
OBI_THRESHOLD_H = 0.10
N_SLICES_H = 50
WAIT_NS_H = 1 * 1e9  # 200ms

df_at_ts = old_df.filter(pl.col("ts_nano") == ts)
lmt = df_at_ts[f"ask_px_09" if SIDE == "BUY" else f"bid_px_09"][0]

ts = old_df.sample(seed=SEED)["ts_nano"][0]
sim = ExecutionSimulator(old_df, latency_ns=LATENCY, replenishment_rate=REPLENISHMENT)
results_h, new_df_h = sim.hybrid_fill(
    SIDE, PARENT_QTY, lmt, ts,
    obi_threshold=OBI_THRESHOLD_H,
    n_slices=N_SLICES_H,
    wait_ns=int(WAIT_NS_H),
)
```

```python

levels = [f"0{i}" for i in range(6)]

bid_sum = pl.sum_horizontal([pl.col(f"bid_sz_{lvl}").cast(pl.Float64) for lvl in levels])
ask_sum = pl.sum_horizontal([pl.col(f"ask_sz_{lvl}").cast(pl.Float64) for lvl in levels])

new_df_h = (
    new_df_h
    .with_columns(
        ((pl.col("ask_px_00") + pl.col("bid_px_00")) / 2).alias("mid")
    )
    .with_columns(
        ((bid_sum - ask_sum) / (bid_sum + ask_sum)).alias("obi")
    )
)
results_h = [r for r in results_h if r['vwap'] != 0.0 and len(r['fills']) != 0]

# Window slice
i_start = results_h[0]['executed_at_idx'] - 100
i_end   = results_h[-1]['executed_at_idx'] + 100

old_window = old_df[i_start:i_end]
new_window = new_df_h[i_start:i_end]

old_ts = pd.to_datetime(old_window["ts_nano"].to_numpy(), unit='ns')
new_ts = pd.to_datetime(new_window["ts_nano"].to_numpy(), unit='ns')

fill_times = pd.to_datetime(
    np.array([r["executed_at_ns"] for r in results_h], dtype=np.int64), unit='ns'
)

fig, ax = plt.subplots(2, 1, figsize=(12, 10))

# ax[0] — Bid/Ask panel
ax[0].plot(old_ts, old_window["ask_px_00"].to_numpy(), color='blue', linestyle='--', label='old ask')
ax[0].plot(old_ts, old_window["bid_px_00"].to_numpy(), color='blue', linestyle='-', label='old bid')
ax[0].plot(new_ts, new_window["ask_px_00"].to_numpy(), color='orange', linestyle='--', label='new ask')
ax[0].plot(new_ts, new_window["bid_px_00"].to_numpy(), color='orange', linestyle='-', label='new bid')
ax[0].vlines(fill_times, ymin=new_window["bid_px_00"].min(), ymax=new_window["ask_px_00"].max(),
             colors='gray', linestyles='dashed', alpha=0.3, label='fills')

# Background shading for OBI regime
obi_vals = new_window["obi"].to_numpy()
ts_vals = new_ts

def regime(v):
    if v > OBI_THRESHOLD_H:
        return 'adverse'
    elif v < -OBI_THRESHOLD_H:
        return 'favorable'
    return None

labeled = {'adverse': False, 'favorable': False}
cur = regime(obi_vals[0]) if len(obi_vals) > 0 else None
start_idx = 0
for i in range(1, len(obi_vals)):
    r = regime(obi_vals[i])
    if r != cur:
        if cur is not None:
            color = 'red' if cur == 'adverse' else 'green'
            lbl = cur if not labeled[cur] else None
            ax[0].axvspan(ts_vals[start_idx], ts_vals[i], color=color, alpha=0.1, label=lbl)
            labeled[cur] = True
        cur = r
        start_idx = i
if cur is not None and len(obi_vals) > 0:
    color = 'red' if cur == 'adverse' else 'green'
    lbl = cur if not labeled[cur] else None
    ax[0].axvspan(ts_vals[start_idx], ts_vals[-1], color=color, alpha=0.1, label=lbl)

ax[0].set_title("Hybrid Fill — Bid/Ask During Execution")

# ax[1] — OBI panel
ax[1].plot(new_ts, new_window["obi"].to_numpy(), marker="o", markersize=2, label='obi')
ax[1].vlines(fill_times, ymin=new_window["obi"].min(), ymax=new_window["obi"].max(),
             colors='gray', linestyles='dashed', alpha=0.3, label='fills')
ax[1].axhline(OBI_THRESHOLD_H, color='red', linestyle='--', label='adverse threshold')
ax[1].axhline(-OBI_THRESHOLD_H, color='green', linestyle='--', label='favorable threshold')
ax[1].set_title("OBI During Execution")

for a in ax:
    a.xaxis.set_major_formatter(mdates.DateFormatter('%H:%M:%S'))
    a.yaxis.set_major_formatter(ScalarFormatter(useOffset=False))
    a.ticklabel_format(style='plain', axis='y')
    a.set_xlabel("Time")
    a.legend()

fig.autofmt_xdate(rotation=45)
plt.tight_layout()
plt.savefig(f'plots/hybrid_fill_{SEED}.png', dpi=300, bbox_inches='tight')
plt.show()

# Summary stats
p0 = old_df.filter(pl.col("ts_nano") == ts)['mid'][0]
filled_results_h = [r for r in results_h if r["vwap"] != 0.0]
total_vwap_h = np.sum([r["vwap"]*r["total_filled"] for r in filled_results_h]) / np.sum([r["total_filled"] for r in filled_results_h])
total_filled_h = np.sum([r["total_filled"] for r in filled_results_h])
mean_child_h = np.mean([r["arrival_cost"] for r in filled_results_h])
median_child_h = np.median([r["arrival_cost"] for r in filled_results_h])
print(f"Total Filled: {total_filled_h}")
print(f"Total Parent Arrival Cost: {total_vwap_h} at arrival cost: {(total_vwap_h - p0)/p0 * 10_000:,.2f} bps")
print(f"Mean Child Arrival Cost: {mean_child_h * 10_000:,.2f} bps")
print(f"Median Child Arrival Cost: {median_child_h * 10_000:,.2f} bps")
```
