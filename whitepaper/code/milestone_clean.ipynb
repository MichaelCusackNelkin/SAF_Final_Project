## Fetch data

```python
import sys
# sys.path.insert(0, '../..')

from data_collection.data_adapters import DatabentoGenericAdapter
from dotenv import load_dotenv
import os

load_dotenv()

adapter = DatabentoGenericAdapter(os.getenv("DATABENTO_API_KEY"))
df, map = await adapter.get_data(
    tickers=["CSCO", "AMZN", "DRH"],
    start_date="2026-03-25",
    end_date="2026-03-31",
    dataset="XNAS.ITCH",
    schema="MBP-10",
    cache_only=True
)
map = {v: k for k,v in map.items()}

print(f"File size (gb): {df.estimated_size('gb'):.2f}")

df.sample(5, seed=42)
```

```python
cols = ["ts_event", "instrument_id"] + [
    col for n in range(0, 10)
    for col in [f"bid_px_{str(n).zfill(2)}", f"ask_px_{str(n).zfill(2)}", f"bid_sz_{str(n).zfill(2)}", f"ask_sz_{str(n).zfill(2)}"]
]
df = df.select(cols)
df.sample(5, seed=42)
```

```python
old_df = df.filter(pl.col("instrument_id") == map["AMZN"]).with_columns(
    pl.col("ts_event").dt.timestamp("ns").alias("ts_nano")
)
old_df = old_df.with_columns(((pl.col("ask_px_00") + pl.col("bid_px_00")) / 2).alias("mid"))
print(f"Sorted: {old_df.sort("ts_nano").equals(old_df)}")
old_df.sample(5, seed=42)
```

```python
LATENCY = 50_000 #50us in ns
REPLENISHMENT = 30.0 #poisson rate of replenishment using wall-clock
PARENT_QTY = 5_000
SEED = 42
```

#### Playing with latency and replenishment

```python
LATENCY = 0
REPLENSIHMENT = 100
sim = ExecutionSimulator(old_df, latency_ns=LATENCY, replenishment_rate=REPLENSIHMENT)
costs = []
for i in range(0, 10):
    ts = old_df.sample(seed=SEED)["ts_nano"][0]
    results, new_df = sim.constant_fill("BUY", PARENT_QTY, ts, wait_ns=1e9, n_chunks=20)
    p0=old_df.filter(pl.col("ts_nano") == ts)['mid'][0]
    filled = np.sum([r["total_filled"] for r in results if r["vwap"] != 0.0])
    total_cost = np.sum([r["vwap"] - p0 for r in results if r["vwap"] != 0.0])
    summed_arrival_cost = np.sum([r["arrival_cost"]for r in results if r["vwap"] != 0.0])
    costs.append({
        "filled": filled,
        "total_cost": total_cost,
        "summed_arrival_cost": summed_arrival_cost
    })
print(LATENCY)
costs = pd.DataFrame(costs)
costs.describe()
```

The result is that this OBI filler is a **hypothetical** at the moment; it gives the result if you had filled at the best OBI opportunities.

Beyond the milestone, we will have to do this sequentially and take into account the in ability to see into the future and know whether another OBI opportunity is coming. Still, this proves that OBI can provide an improvement over constant or naive filling.
