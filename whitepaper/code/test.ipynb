```python
from data_collection.data_adapters import DatabentoGenericAdapter
from dotenv import load_dotenv
import os
import polars as pl

load_dotenv()

adapter = DatabentoGenericAdapter(os.getenv("DATABENTO_API_KEY"))
raw, map = await adapter.get_data(
    tickers=["CSCO", "AMZN", "DRH"],
    start_date="2026-03-25",
    end_date="2026-03-27",
    dataset="XNAS.ITCH",
    schema="MBP-10",
    cache_only=False
)
map = {v: k for k,v in map.items()}

print(f"File size (gb): {raw.estimated_size('gb'):.2f}")

raw.sample(5, seed=42)
```

```python
TICKER = "DRH"

# Get Data and select to relavent columns
cols = ["ts_event", "instrument_id", "action", "price", "size"] + [
    col for n in range(0, 10)
    for col in [f"bid_px_{str(n).zfill(2)}", f"ask_px_{str(n).zfill(2)}", f"bid_sz_{str(n).zfill(2)}", f"ask_sz_{str(n).zfill(2)}"]
]
df = raw.select(cols)
df = df.filter(pl.col("instrument_id") == map[TICKER]).with_columns(
    pl.col("ts_event").dt.timestamp("ns").alias("ts_nano")
)
df = df.with_columns(((pl.col("ask_px_00") + pl.col("bid_px_00")) / 2).alias("mid"))
print(f"Sorted: {df.sort("ts_nano").equals(df)}")
print(f"Estimated Data Size: {df.estimated_size('gb'):.2f} GB")
```

## loading pairs

```python
import json
import sys
from pathlib import Path

sys.path.insert(0, str(Path(".")))
from trials import _trial_cache_key

CACHE_DIR = Path("../report/plots_and_data/section_3/successful_run/cost_data/")

def load_pair(hybrid_parquet: Path):
    """Given a hybrid parquet path, load it and its matching constant parquet."""
    params_path = CACHE_DIR / f"{hybrid_parquet.stem}.json"
    with open(params_path) as f:
        params = json.load(f)

    const_key = _trial_cache_key(**{**params, "method": "constant"})
    const_path = CACHE_DIR / f"{const_key}.parquet"

    hybrid_df = pl.read_parquet(hybrid_parquet)
    const_df = pl.read_parquet(const_path)
    return hybrid_df, const_df, params

# --- pick a hybrid parquet to inspect ---
hybrid_path = CACHE_DIR / "204b0a2f144401b56ae861cbac7ec2a6.parquet"

hybrid_df, const_df, params = load_pair(hybrid_path)

print("Params:", params)
print("\nHybrid shape:", hybrid_df.shape)
print("Constant shape:", const_df.shape)
```

```python
hybrid_df
```

```python
from scipy import stats
t_stat, p_val = stats.ttest_ind(
    const_df['parent_arrival_cost'],
    hybrid_df['parent_arrival_cost'],
    equal_var=False
)
print(f"T-stat: {t_stat:,.2f}")
print(f"P-value significance: {p_val:,.2f}")
```

```python
from matplotlib import pyplot as plt
plot_hybrid = hybrid_df.with_columns(pl.from_epoch("ts_event", time_unit="ns").alias("ts_event").dt.convert_time_zone('US/Eastern'))
plot_constant = const_df.with_columns(pl.from_epoch("ts_event", time_unit="ns", ).alias("ts_event").dt.convert_time_zone('US/Eastern'))

# -- Parent Costs --
log = True
plt.hist(plot_hybrid["parent_arrival_cost"], bins=25, edgecolor="black", color="tab:orange", alpha=0.7, label="Hybrid", log=log)
plt.hist(plot_constant["parent_arrival_cost"], bins=25, edgecolor="black", color="tab:blue", alpha=0.7, label="Naive", log=log)
plt.axvline(plot_hybrid['parent_arrival_cost'].mean(), color='tab:orange', linestyle='--', linewidth=1, label='hybrid mean')
plt.axvline(plot_constant['parent_arrival_cost'].mean(), color='tab:blue', linestyle='--', linewidth=1, label='naive mean')
plt.xlabel("Cost per Share (bps)")
plt.ylabel("Count")
plt.title("Parent Cost Per Share Comparison")
plt.legend()
plt.show()
```
